```python
import boto3
import json
```

```python
bedrock_runtime = boto3.client('bedrock-runtime', region_name='us-east-1')
```

```python
# Construct request body adhering to Nova schema requirements
request_payload = {
    "system": [{"text": "Be concise and use clear formatting."}],
    "messages": [
        {
            "role": "user",
            "content": [{"text": "Hi tell me about making a good PPT for marketing"}]
        }
    ],
    "inferenceConfig": {
        "maxTokens": 500,
        "temperature": 0.3
    }
}
```

```python
# Invoke using Cross-Region Inference or standard endpoint IDs
response = bedrock_runtime.invoke_model(
    modelId="us.amazon.nova-lite-v1:0", 
    body=json.dumps(request_payload)
)
```

```python
# Parse response matching Nova output schema layout
response_body = json.loads(response.get('body').read())
output_text = response_body["output"]["message"]["content"][0]["text"]
```
